```python
import polars as pl
from airports.airport_lookup import AirportLookup

airport_lookup = AirportLookup()


def canonical_airport_ident(ident: str | None) -> str | None:
    """Resolve ICAO, GPS, FAA/local, or IATA codes to one OurAirports ident."""
    if not ident:
        return None

    ident = ident.strip().upper()

    if ident in airport_lookup.ident_index:
        return ident

    for index in (
        airport_lookup.gps_code_to_ident_index,
        airport_lookup.local_code_to_ident_index,
        airport_lookup.iata_to_ident_index,
    ):
        canonical = index.get(ident)
        if canonical is not None:
            return canonical

    return None


icaos_all_both = (
    df_flights.group_by("icao")
    .agg((pl.col("match_status") == "both").all().alias("all_both"))
    .filter("all_both")
    .get_column("icao")
)

takeoff_disagreements = (
    df_flights
    .filter(
        ~pl.col("same_takeoff_airport_ident")
        & (pl.col("match_status") == "both")
        & pl.col("icao").is_in(icaos_all_both.implode())
    )
    .with_columns(
        pl.col("takeoff_airport_ident")
        .map_elements(canonical_airport_ident, return_dtype=pl.String)
        .alias("test_canonical_airport"),
        pl.col("takeoff_airport_ident_df1")
        .map_elements(canonical_airport_ident, return_dtype=pl.String)
        .alias("gold_canonical_airport"),
    )
    .with_columns(
        pl.when(
            pl.col("test_canonical_airport").is_null()
            | pl.col("gold_canonical_airport").is_null()
        )
        .then(pl.lit("unresolved"))
        .when(
            pl.col("test_canonical_airport")
            == pl.col("gold_canonical_airport")
        )
        .then(pl.lit("identifier_alias"))
        .otherwise(pl.lit("genuine_disagreement"))
        .alias("disagreement_type")
    )
)

genuine_takeoff_disagreements = (
    takeoff_disagreements
    .filter(pl.col("disagreement_type") == "genuine_disagreement")
    .select(
        "icao",
        "callsign",
        "takeoff_time",
        "takeoff_time_df1",
        "takeoff_airport_ident",
        "takeoff_airport_ident_df1",
        "test_canonical_airport",
        "gold_canonical_airport",
    )
)

genuine_takeoff_disagreements
```

## Attribution of non-`both` flight-segmentation rows

Re-running the evaluation with the current code/data gives **32 unmatched rows across 17 ICAOs** (12 `df0_only`, 20 `df1_only`). I checked each aircraft against both the raw ADSB.lol track used by `df0` and the independent ADS-B Exchange track used to qualify SFDPS.

Primary attribution: **27 rows / 15 ICAOs are caused by `df0`**, while **5 rows / 2 ICAOs are caused by `df1`**. The `a25ec5` result is policy-sensitive: SFDPS records a landing at KACT and another takeoff four minutes later, so it is two legs under the current takeoff-to-landing definition; it could instead be treated as one sortie if touch-and-goes should not split flights.

Important: `match_status` labels which side was unmatched, not which side was correct. For example, a merged `df0` flight usually produces one `df0_only` row plus two `df1_only` rows, even though all three rows have the same `df0` root cause.

```python
MISMATCH_REVIEW = [
    {"icao": "a00128", "attribution": "df0_wrong", "reason": "ADSB.lol misses KELD-KLIT; ADS-B Exchange confirms the SFDPS leg"},
    {"icao": "a003a8", "attribution": "df0_wrong", "reason": "df0 merges KBCT-MYEN and MYEN-KFXE across the MYEN stop"},
    {"icao": "a06abb", "attribution": "df0_wrong", "reason": "ADSB.lol misses the early KTPA-KFLL leg; ADS-B Exchange confirms it"},
    {"icao": "a10b6b", "attribution": "df0_wrong", "reason": "KCRQ-KFUL is present in raw ADSB.lol but omitted from df0"},
    {"icao": "a25ec5", "attribution": "df0_wrong", "reason": "df0 merges KFTW-KACT and KACT-KFTW; four-minute KACT turnaround (touch-and-go policy-sensitive)"},
    {"icao": "a2fa4d", "attribution": "df0_wrong", "reason": "second KBZN local flight exists in both ADS-B feeds but is omitted from df0"},
    {"icao": "a32756", "attribution": "df1_wrong", "reason": "three real low-altitude KRIC local sorties appear in both ADS-B feeds but are absent from SFDPS"},
    {"icao": "a5536c", "attribution": "df0_wrong", "reason": "ADSB.lol begins mid-flight at 17:26; ADS-B Exchange confirms the TXKF departure near 15:39"},
    {"icao": "a559de", "attribution": "df0_wrong", "reason": "a stray midnight cluster is merged with the actual 13:29 KBHM departure"},
    {"icao": "a6ea99", "attribution": "df1_wrong", "reason": "SFDPS says 15:30, but ADS-B Exchange still shows the aircraft on the ground at KGYY at 16:20; df0's 16:36 departure is supported"},
    {"icao": "a76ffa", "attribution": "df0_wrong", "reason": "df0 merges KSAT-KELP and KELP-KSBA across the El Paso stop"},
    {"icao": "a83bc6", "attribution": "df0_wrong", "reason": "middle KART-KRIC leg is missing from ADSB.lol/df0 but confirmed by ADS-B Exchange"},
    {"icao": "a95169", "attribution": "df0_wrong", "reason": "df0 merges KAPA-KASE and KASE-KAUS across the Aspen stop"},
    {"icao": "a953e4", "attribution": "df0_wrong", "reason": "df0 merges KVRB-KRST and KRST-KANE across the Rochester stop"},
    {"icao": "ab14c0", "attribution": "df0_wrong", "reason": "midnight carryover plus low-altitude coverage gaps split the SFDPS KJQF round-robin"},
    {"icao": "ab236f", "attribution": "df0_wrong", "reason": "df0 merges KGKY-KABI and KABI-KGKY across the seven-minute KABI stop"},
    {"icao": "ac97c2", "attribution": "df0_wrong", "reason": "low-altitude coverage gaps split one KJYO local round-trip and create a spurious extra df0 segment"},
]

mismatch_counts = (
    df_flights.filter(pl.col("match_status") != "both")
    .group_by("icao")
    .agg(
        pl.len().alias("unmatched_rows"),
        (pl.col("match_status") == "df0_only").sum().alias("df0_only_rows"),
        (pl.col("match_status") == "df1_only").sum().alias("df1_only_rows"),
    )
)
mismatch_review = pl.DataFrame(MISMATCH_REVIEW).join(mismatch_counts, on="icao", how="left")
assert set(mismatch_counts["icao"]) == set(mismatch_review["icao"]), "Review is stale; investigate new/removed mismatches"
mismatch_review.sort(["attribution", "icao"])
```
